# Categorical Analysis 

```python
## Creating income bands, calculating the number of individuals in each income band, and visualizing the results
bins = [0, 100000, 300000, float('inf')]
labels = ["Low", "Med", "High"]
df["Income Band"] = pd.cut(
    df["Estimated Income"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

df["Income Band"].value_counts().plot(kind="bar")
```

```python
# Examining the distribution of unique categories in categorical columns
categorical_cols = df[["Amount of Credit Cards", "BRId", "GenderId", "IAId", "Nationality", "Occupation", "Fee Structure", "Loyalty Classification", "Properties Owned", "Risk Weighting", "Income Band"]]
for col in categorical_cols:
    print(f"value counts for '{col}':")
    display(df[col].value_counts())
```

```python
# Univariate visualization of the distribution of unique categories in categorical columns
for i, predictor in enumerate(categorical_cols):
    plt.figure(figsize=(5, 5))
    
    sns.countplot(data=df, x=predictor)
    
    plt.title(f"Distribution of {predictor}")
    plt.xlabel(predictor)
    plt.ylabel("Count")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

```python
# Bivariate visualization of the distribution of unique categories in categorical columns
for i, predictor in enumerate(categorical_cols):
    plt.figure(figsize=(5, 5))
    
    sns.countplot(data=df, x=predictor, hue="GenderId")
    
    plt.title(f"Distribution of {predictor} by Gender")
    plt.xlabel(predictor)
    plt.ylabel("Count")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

```

```python
# Bivariate visualization of the distribution of unique categories in categorical columns

for i, predictor in enumerate(categorical_cols):
    plt.figure(figsize=(5, 5))
    sns.countplot(data=df, x=predictor, hue="Nationality")
    
    plt.title(f"Distribution of {predictor} by Gender")
    plt.xlabel(predictor)
    plt.ylabel("Count")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

```

## Numerical Analysis

```python
numerical_cols= ["Age", "Estimated Income",	"Superannuation Savings", "Credit Card Balance", "Bank Loans", "Bank Deposits", "Checking Accounts", "Saving Accounts", "Foreign Currency Account", "Business Lending"]

# univariate analysis and visualization 
plt.figure(figsize=(15,80))
for i, col in enumerate(numerical_cols):
    plt.subplot(11,1,i+1)
    sns.histplot(df[col], kde=True)
    plt.title(col)
plt.show()
```

```python
# heatmaps

correlation_matrix = df[numerical_cols].corr()
plt.figure(figsize=(12,12))
sns.heatmap(correlation_matrix, annot=True, cmap='crest',  fmt='.2f')
plt.title('Correlation Heatmap of Numerical Variables')
plt.tight_layout()
plt.show()
```
